```python
from image_parameter_script import compute_image_parameters
import csv

pvals = [0.0, 0.5, 0.75, 1.0]
nvals = [0.0, 0.5, 1.0, 1.5]
polarization = False
BLUR = [1, 5, 10, 20]
if polarization:
    csv_filename = "./Images/image_parameter_analysis_polar.csv"
else: 
    csv_filename = "./Images/image_parameter_analysis.csv"

with open(csv_filename, mode="w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["blur","p_val", "n_I", "d", "w", "eta", "Asym", "argbeta2"])
    for blur in BLUR:
        for p in pvals:
            for n in nvals:
                if polarization:
                    path_to_fitsfile = f"./Images/m87_model_pval{p}_n{n}_polar.fits"
                else:
                    path_to_fitsfile = f"./Images/m87_model_pval{p}_n{n}.fits"
                d, w, eta, Asym, argbeta2 = compute_image_parameters(path_to_fitsfile, blur_kernel_uas=blur)
                writer.writerow([blur,p, n, d, w, eta, Asym, argbeta2])
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import numpy as np
from matplotlib.ticker import MultipleLocator, FuncFormatter

df = pd.read_csv("./Images/image_parameter_analysis.csv")

stats = ["eta", "argbeta2"]

for blur, df_blur in df.groupby("blur"):
    for stat in stats:
        fig, ax = plt.subplots(figsize=(8, 5))
        for n_I, df_nI in df_blur.groupby("n_I"):
            ax.plot(df_nI["p_val"], df_nI[stat], marker="o", label=f"n_I={n_I}")
        if stat == "eta": 
            scale = 32
        else:
            scale = 8
        ax.set_title(r"$\theta_B$ not in emissivity", fontsize = 20)
        ax.set_xlabel("p", fontsize = 20, style='italic')
        ax.set_ylabel(stat, fontsize = 20, style='italic')
        ax.legend()
        ax.grid(alpha=0.3)
        ax.yaxis.set_major_locator(MultipleLocator(np.pi/scale))  
        ax.yaxis.set_major_formatter(FuncFormatter(lambda x, _: rf"${x/np.pi:.2g}\pi$"))
        if blur == 5:
            fig.savefig(os.path.join("./Images", f"final_fixed/{stat}vp_{blur}.png"))
            plt.show() 
        plt.close(fig)    
```
